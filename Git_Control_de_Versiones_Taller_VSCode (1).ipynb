````python
# ============================================================
#  ARRANQUE DEL CUADERNO
#  Ejecute esta celda cada vez que abra el cuaderno o reinicie el kernel.
#  No necesita modificarla.
# ============================================================
import os
import platform
import re
import shutil
import subprocess
import sys
from pathlib import Path

from IPython.core.magic import register_cell_magic

# --- 1) Carpeta de trabajo: la misma donde está guardado este cuaderno ---
_archivo_cuaderno = globals().get("__vsc_ipynb_file__")
CARPETA_CUADERNO = Path(_archivo_cuaderno).resolve().parent if _archivo_cuaderno else Path.cwd()
os.chdir(CARPETA_CUADERNO)
CARPETA_TALLER = CARPETA_CUADERNO / "taller_git"
CARPETA_TALLER.mkdir(exist_ok=True)

# --- 2) Entorno propio de Git (no toca su configuración real) ---
CONFIG_GIT = CARPETA_TALLER / ".gitconfig_taller"
os.environ["GIT_CONFIG_GLOBAL"] = str(CONFIG_GIT)  # la "configuración global" de este cuaderno
os.environ["GIT_TERMINAL_PROMPT"] = "0"            # Git nunca se queda esperando una contraseña
os.environ["GIT_EDITOR"] = "true"                  # Git nunca abre un editor de texto
os.environ["GIT_MERGE_AUTOEDIT"] = "no"
os.environ["LC_ALL"] = "C"                         # mensajes de Git siempre en inglés
os.environ["LANGUAGE"] = "en"

_marcas = []


def _anotar(descripcion, ok, ayuda=""):
    _marcas.append(ok)
    print(("✅ " if ok else "❌ ") + descripcion)
    if not ok and ayuda:
        print("   → " + ayuda)
    return ok


# --- 3) Python ---
_anotar("Python %s (se pide 3.9 o superior)" % platform.python_version(),
        sys.version_info >= (3, 9),
        "Instale un Python más reciente (python.org) y selecciónelo como kernel.")

# --- 4) Git ---
GIT = shutil.which("git")
_version_git = None
_texto_version = ""
if GIT:
    _texto_version = subprocess.run([GIT, "--version"], capture_output=True, text=True,
                                    encoding="utf-8", errors="replace").stdout.strip()
    _m = re.search(r"(\d+)\.(\d+)", _texto_version)
    if _m:
        _version_git = (int(_m.group(1)), int(_m.group(2)))
_anotar("Git está instalado" + (" (%s)" % _texto_version if GIT else ""), bool(GIT),
        "Instale Git (vea la sección 0.1) y luego cierre y vuelva a abrir VS Code.")
_git_nuevo = bool(_version_git and _version_git >= (2, 32))
_anotar("Git es 2.32 o superior", _git_nuevo,
        "Actualice Git: su versión es demasiado antigua para este cuaderno.")


# --- 5) Terminal Bash (en Windows: Git Bash) ---
def _buscar_bash():
    if platform.system() == "Windows":
        candidatos = []
        if GIT:
            for carpeta in list(Path(GIT).resolve().parents)[:4]:
                candidatos.append(carpeta / "bin" / "bash.exe")
        for variable in ("ProgramFiles", "ProgramFiles(x86)", "LOCALAPPDATA"):
            base = os.environ.get(variable)
            if base:
                candidatos.append(Path(base) / "Git" / "bin" / "bash.exe")
                candidatos.append(Path(base) / "Programs" / "Git" / "bin" / "bash.exe")
        for candidato in candidatos:
            if candidato.is_file():
                return str(candidato)
        return None
    return shutil.which("bash")


BASH = _buscar_bash()
_anotar("Terminal Bash encontrada" + (" (%s)" % BASH if BASH else ""), bool(BASH),
        "En Windows, Bash viene con Git for Windows (Git Bash): reinstale Git y reabra VS Code."
        if platform.system() == "Windows" else
        "Bash viene de serie en Mac y Linux; instálelo con el gestor de paquetes de su sistema.")

# En Windows se antepone la carpeta de herramientas de Git Bash (sed, grep, find...) para que
# no se confundan con programas del mismo nombre que trae Windows.
_RUTAS_BASH = []
if BASH and platform.system() == "Windows":
    _raiz_git = Path(BASH).resolve().parent.parent
    _RUTAS_BASH = [str(p) for p in (_raiz_git / "usr" / "bin", _raiz_git / "mingw64" / "bin") if p.is_dir()]

# Ayuda propia del cuaderno: cambia un texto por otro dentro de un archivo.
# (Funciona igual en Windows, Mac y Linux.)
PRELUDIO_BASH = """
reemplazar() {
    sed -i.bak "s/$1/$2/g" "$3" && rm -f "$3.bak"
}
"""


def _correr_bash(guion, carpeta=None):
    """Ejecuta un guion en Bash y devuelve (código_de_salida, texto_de_salida)."""
    datos = (PRELUDIO_BASH + guion).replace("\r\n", "\n").encode("utf-8")
    entorno = os.environ.copy()
    if _RUTAS_BASH:
        entorno["PATH"] = os.pathsep.join(_RUTAS_BASH + [entorno.get("PATH", "")])
    r = subprocess.run([BASH], input=datos, stdout=subprocess.PIPE, stderr=subprocess.STDOUT,
                       cwd=str(carpeta or os.getcwd()), env=entorno)
    return r.returncode, r.stdout.decode("utf-8", errors="replace").replace("\r\n", "\n")


if BASH:
    _prueba = (
        "set -e\n"
        "printf 'hola\\n' > _prueba.txt\n"
        "reemplazar hola adios _prueba.txt\n"
        "grep -q adios _prueba.txt\n"
        "cat _prueba.txt | head -n 1 | wc -l > /dev/null\n"
        "tail -n 1 _prueba.txt > /dev/null\n"
        "ls > /dev/null\n"
        "find . -maxdepth 1 -name _prueba.txt | grep -q _prueba\n"
        "rm -f _prueba.txt _prueba.txt.bak\n"
        "echo LISTO\n"
    )
    _codigo, _salida = _correr_bash(_prueba, CARPETA_TALLER)
    _anotar("Las herramientas de terminal (sed, grep, find, cat...) responden",
            _codigo == 0 and _salida.strip().endswith("LISTO"),
            "Salida recibida: " + _salida.strip()[:200])

    @register_cell_magic("bash")
    def _celda_bash(linea, celda):
        """Reemplaza a %%bash: usa la terminal encontrada arriba (Git Bash en Windows)."""
        codigo, salida = _correr_bash(celda)
        if salida:
            print(salida, end="" if salida.endswith("\n") else "\n")
        if codigo != 0:
            print("\n⚠️ El último comando de la celda terminó con error (código %d)." % codigo)

# --- 6) Configuración de Git propia del cuaderno ---
if GIT and _git_nuevo:
    def _config(*argumentos):
        return subprocess.run([GIT, "config", "--global", *argumentos], capture_output=True,
                              text=True, encoding="utf-8", errors="replace")

    for _clave, _valor in [("init.defaultBranch", "main"),    # la rama principal se llama main
                           ("pull.rebase", "false"),          # git pull usa la estrategia de fusión
                           ("core.autocrlf", "false")]:       # sin avisos de saltos de línea
        _config(_clave, _valor)
    if not _config("--get", "user.name").stdout.strip():
        _config("user.name", "Aprendiz SENA")
    if not _config("--get", "user.email").stdout.strip():
        _config("user.email", "aprendiz@ejemplo.com")
    _anotar("Configuración propia de Git creada (su configuración real no se modifica)",
            CONFIG_GIT.is_file(), "No se pudo escribir en " + str(CONFIG_GIT))

# --- 7) Carpeta de trabajo ---
try:
    _tmp = CARPETA_TALLER / "_escritura.tmp"
    _tmp.write_text("ok", encoding="utf-8")
    _tmp.unlink()
    _anotar("Se puede escribir en la carpeta de trabajo", True)
except OSError as error:
    _anotar("Se puede escribir en la carpeta de trabajo", False,
            "Mueva el cuaderno a una carpeta suya, por ejemplo en Documentos. (%s)" % error)

# --- 8) Advertencia (no bloquea): la carpeta no debería estar dentro de otro repositorio ---
if GIT:
    _dentro = subprocess.run([GIT, "-C", str(CARPETA_CUADERNO), "rev-parse", "--is-inside-work-tree"],
                             capture_output=True, text=True).stdout.strip() == "true"
    if _dentro:
        print("⚠️ Este cuaderno está dentro de otro repositorio de Git. Funcionará, pero conviene "
              "moverlo a una carpeta que no sea un repositorio, para no mezclar sus prácticas con un proyecto real.")
    else:
        print("✅ La carpeta del cuaderno no está dentro de otro repositorio")

print()
print("Carpeta de trabajo:", CARPETA_CUADERNO)
if not all(_marcas):
    raise RuntimeError("El entorno todavía no está listo: revise los ❌ de arriba, corrija y vuelva a ejecutar esta celda.")
print("🎉 Entorno listo. Puede continuar con la sección 0.3.")
````

## 0.4 Ayudas para las verificaciones

Esta celda define dos pequeñas funciones que usaremos en las celdas de verificación automática:

- `git_texto(carpeta, ...)`: ejecuta un comando de Git en una carpeta y devuelve su resultado como texto.
- `comprobar(descripcion, condicion)`: imprime ✅ si la condición se cumple y ❌ si no.

No necesita modificarlas, pero puede leerlas: son Python sencillo.

````python
import subprocess

def git_texto(carpeta, *argumentos):
    """Ejecuta 'git -C carpeta ...' y devuelve la salida como texto ("" si hubo error)."""
    r = subprocess.run(["git", "-C", carpeta, *argumentos],
                       capture_output=True, text=True, encoding="utf-8", errors="replace")
    return r.stdout.strip() if r.returncode == 0 else ""

def comprobar(descripcion, condicion):
    """Imprime una marca ✅ o ❌ y devuelve la condición."""
    print(("✅ " if condicion else "❌ ") + descripcion)
    return condicion

print("Ayudas listas.")
````

# Parte 1. ¿Qué es Git? Git frente a GitHub

## 1.1 El problema que Git resuelve

Seguramente ha visto (o ha hecho) algo así con un trabajo escrito:

```
informe.docx
informe_final.docx
informe_final_v2.docx
informe_final_v2_CORREGIDO.docx
informe_final_v2_CORREGIDO_ahora_si.docx
```

Cada copia ocupa espacio, nadie sabe cuál es la buena y, si dos personas editan a la vez, alguien termina pisando el trabajo de otro. Un **sistema de control de versiones** resuelve esto: usted mantiene un solo proyecto y el sistema recuerda cada versión importante, con fecha, autor y una nota explicando el cambio.

**Git** es el sistema de control de versiones más usado del mundo. Fue creado en 2005 por Linus Torvalds, el creador de Linux, para coordinar a miles de programadores.

## 1.2 Git y GitHub no son lo mismo

Una analogía: **Git es como el programa de correo en su computador**, y **GitHub es como el servicio de correo en internet** donde se guardan y comparten los mensajes. Usted puede usar Git sin GitHub, pero no GitHub sin Git.

| | Git | GitHub |
|---|---|---|
| ¿Qué es? | Un programa que se instala en su computador | Un sitio web (de la empresa Microsoft) |
| ¿Para qué sirve? | Guardar el historial de cambios de un proyecto | Publicar repositorios, compartirlos y colaborar |
| ¿Necesita internet? | No, funciona sin conexión | Sí |
| ¿Quién lo usa? | Cada desarrollador, en su máquina | Equipos que comparten código |
| Funciones propias | `commit`, `branch`, `merge`, `log`... | Pull Requests, Issues, perfiles, páginas web |
| Alternativas | Mercurial, Subversion | GitLab, Bitbucket |

## 1.3 Vocabulario básico

| Palabra | Significado sencillo |
|---|---|
| **Repositorio** (repo) | La carpeta de su proyecto más su historial de cambios |
| **Commit** | Una "foto" del proyecto en un momento dado, con un mensaje que explica el cambio |
| **Rama** (branch) | Una línea de trabajo paralela, para probar algo sin dañar lo que ya funciona |
| **Remoto** (remote) | Una copia del repositorio en otro lugar (por ejemplo, GitHub) |
| **Clonar** | Descargar una copia completa de un repositorio, con todo su historial |
| **Push / Pull** | Subir sus cambios al remoto / traer los cambios de los demás |

## 2.1 La analogía de la mudanza

Imagine que se muda de casa:

- **La sala de su casa** es el **directorio de trabajo**: aquí están sus archivos y aquí los edita.
- **La caja de mudanza** es el **área de preparación** (*staging area*): usted decide qué cosas del desorden de la sala van a la caja. Todavía nada ha salido de la casa.
- **El camión con la caja sellada y etiquetada** es el **repositorio** (el historial): cuando usted cierra la caja y le pone una etiqueta ("mensaje del commit"), esa versión queda guardada.

```
 Directorio de trabajo        Área de preparación            Repositorio
 (sus archivos, donde         (staging / index:              (carpeta .git:
  usted edita)                 lo que irá en el próximo       el historial de
                               commit)                        commits)
       │                              │                             │
       │ ──────── git add ───────────▶│                             │
       │                              │ ───── git commit ─────────▶ │
       │ ◀────────── git restore ──── │                             │
```

¿Por qué existe la caja intermedia? Porque así usted puede elegir **qué cambios van juntos** en un commit, aunque haya tocado diez archivos distintos.

## 2.2 El comando que lo muestra todo: `git status`

La versión corta, `git status --short`, muestra dos columnas de letras: la **izquierda** describe el área de preparación y la **derecha** el directorio de trabajo.

| Lo que ve | Qué significa |
|---|---|
| `?? archivo` | Archivo nuevo que Git todavía no sigue |
| `A  archivo` | Archivo nuevo ya preparado para el próximo commit |
| ` M archivo` | Modificado en el directorio de trabajo, sin preparar |
| `M  archivo` | Modificado y preparado |
| `MM archivo` | Preparado, pero usted lo volvió a modificar después |

Veámoslo con un ejemplo real. Observe cómo cambia la salida de `git status --short` en cada paso.

````python
%%bash
rm -rf taller_git/s02; mkdir -p taller_git/s02; cd taller_git/s02
git init -q -b main
echo "# TiendaTech Cali" > README.md
echo "--- Estado 1: archivo nuevo; Git todavía no lo sigue ---"
git status --short
git add README.md
echo "--- Estado 2: el archivo está en la caja (área de preparación) ---"
git status --short
git commit -q -m "docs: crear README"
echo "--- Estado 3: guardado en el historial (no queda nada pendiente) ---"
git status --short
git log --oneline
````

# Parte 3. Git por dentro (sin magia)

Antes de aprender más comandos conviene mirar qué hay realmente dentro de un repositorio. Verá que Git no tiene nada misterioso: es una carpeta con archivos y una idea muy sencilla.

> **Idea central:** un commit no guarda "los cambios"; guarda **una foto completa** del proyecto. Git es una base de datos donde cada objeto se identifica por una **huella digital** (un *hash* de 40 caracteres) calculada a partir de su contenido.

## 3.1 La carpeta `.git`

Cuando usted ejecuta `git init`, Git crea una carpeta oculta llamada `.git`. **Esa carpeta es el repositorio.** Si la borra, el proyecto queda sin historial; si la copia, copia toda la historia.

> 💡 En VS Code, la carpeta `.git` no aparece en el panel de archivos porque está oculta; por eso aquí la miramos con comandos.

````python
%%bash
rm -rf taller_git/s03; mkdir -p taller_git/s03; cd taller_git/s03
git init -q -b main

echo "--- Contenido de la carpeta .git recién creada ---"
ls .git
echo
echo "--- HEAD: ¿en qué rama estamos? (es un archivo de texto) ---"
cat .git/HEAD
````

## 3.2 Un archivo se convierte en un objeto ("blob")

Cuando usted prepara un archivo con `git add`, Git guarda su contenido como un objeto llamado **blob** y lo nombra con su huella digital. El nombre depende **solo del contenido**: dos archivos con el mismo texto tienen la misma huella.

Para demostrar que no hay magia, calcularemos la huella nosotros mismos con Python y la compararemos con la que calcula Git. La receta de Git es: tomar la palabra `blob`, un espacio, el largo del contenido, un byte nulo, y luego el contenido; con todo eso se calcula el algoritmo SHA-1.

````python
import hashlib
import subprocess

contenido = "hola\n".encode("utf-8")
cabecera = b"blob " + str(len(contenido)).encode() + b"\x00"
hash_python = hashlib.sha1(cabecera + contenido).hexdigest()

hash_git = subprocess.run(["git", "hash-object", "--stdin"], input=contenido,
                          capture_output=True).stdout.decode().strip()

print("Calculado con Python:", hash_python)
print("Calculado por Git   :", hash_git)
print("¿Son iguales?       :", hash_python == hash_git)
````

## 3.3 Armar un commit "a mano"

Los comandos que usted usa a diario (`add`, `commit`) son atajos de comandos más básicos. Aquí los ejecutamos uno por uno para ver las piezas:

1. `git add` → guarda el *blob* y lo registra en la caja.
2. `git write-tree` → crea un objeto **árbol** (*tree*): la lista de archivos de la foto.
3. `git commit-tree` → crea el objeto **commit**: apunta al árbol, dice quién y cuándo, y trae el mensaje.
4. `git update-ref` → mueve la rama para que apunte al nuevo commit.

````python
%%bash
cd taller_git/s03

echo "producto,precio" > catalogo.csv
echo "Laptop Gamer,4500000" >> catalogo.csv

git add catalogo.csv                                              # 1) crea el blob
echo "--- Objetos guardados hasta ahora (solo el blob) ---"
find .git/objects -type f

ARBOL=$(git write-tree)                                           # 2) crea el árbol
COMMIT=$(git commit-tree "$ARBOL" -m "commit hecho a mano")       # 3) crea el commit
git update-ref refs/heads/main "$COMMIT"                          # 4) mueve la rama main

echo "--- Objetos ahora: blob + árbol + commit ---"
find .git/objects -type f | wc -l

echo "--- Lo que 'git log' ve gracias a ese commit ---"
git log --oneline

echo "--- El commit por dentro ---"
git cat-file -p HEAD
echo "--- El árbol por dentro ---"
git cat-file -p "HEAD^{tree}"
echo "--- El blob por dentro ---"
git cat-file -p "$(git rev-parse HEAD:catalogo.csv)"
````

Observe lo que acaba de ver:

- El **commit** contiene: la huella del árbol (`tree ...`), el autor, el momento y el mensaje.
- El **árbol** contiene: el nombre de cada archivo y la huella de su blob.
- El **blob** contiene: el texto del archivo, sin más.

## 3.4 Una rama es solo un archivo con una huella

¿Qué es entonces una rama? Mire dentro de `.git/refs/heads/`.

````python
%%bash
cd taller_git/s03
echo "--- Archivos de ramas ---"
ls .git/refs/heads
echo "--- Contenido del archivo de la rama main (la huella del último commit) ---"
cat .git/refs/heads/main
echo "--- ¿Coincide con lo que Git calcula? ---"
git rev-parse main
````

Una rama es un archivo de 41 caracteres (la huella de un commit). **Crear una rama es crear un archivo**; por eso las ramas en Git son tan baratas y rápidas. Y `HEAD` es simplemente el archivo que dice cuál rama tiene usted activa.

## 4.2 Cómo escribir un buen mensaje de commit

Dentro de seis meses, usted (o su compañero) leerá estos mensajes para entender qué pasó. Un mensaje como `arreglo` o `cambios` no ayuda a nadie. Una convención muy extendida es escribir **el tipo de cambio, dos puntos y una frase corta en presente**:

```
tipo: descripción corta de lo que hace el cambio
```

| Tipo | Úselo cuando... | Ejemplo |
|---|---|---|
| `feat` | agrega una función o contenido nuevo | `feat: agregar teclado mecánico al catálogo` |
| `fix` | corrige un error | `fix: corregir precio del mouse` |
| `docs` | cambia solo documentación | `docs: crear README` |

| `test` | agrega o corrige pruebas | `test: probar descuento del 10 %` |
| `chore` | tareas de mantenimiento (configuración, `.gitignore`...) | `chore: agregar .gitignore` |

Reglas prácticas: **una idea por commit**, máximo unos 60 caracteres en la primera línea, y que la frase pueda completar *"Este commit..."*: "Este commit agrega el teclado mecánico".

# Parte 5. Ver cambios con `diff`

Antes de guardar un cambio conviene **revisarlo**. El comando `git diff` compara versiones y muestra, línea por línea, qué se quitó y qué se agregó.

| Comando | Compara... |
|---|---|
| `git diff` | directorio de trabajo contra el área de preparación (lo que **aún no ha preparado**) |
| `git diff --staged` | área de preparación contra el último commit (lo que **va a guardar**) |
| `git diff HEAD~1 HEAD` | el penúltimo commit contra el último |
| `git diff --stat` | solo un resumen: cuántas líneas cambiaron por archivo |

**Cómo leer un diff:** las líneas que empiezan con `-` son las que **se quitaron**; las que empiezan con `+` son las que **se agregaron**; la línea `@@ ... @@` indica en qué parte del archivo ocurrió el cambio. (En VS Code, el panel de Control de código fuente muestra estos mismos cambios con colores; lo veremos en la Parte 12.)

````python
%%bash
rm -rf taller_git/s05; mkdir -p taller_git/s05; cd taller_git/s05
git init -q -b main
printf "producto,categoria,precio\nMouse inalámbrico,Perifericos,55000\nTeclado mecánico,Perifericos,180000\n" > catalogo.csv
git add catalogo.csv
git commit -q -m "feat: agregar catálogo inicial"

reemplazar 55000 59000 catalogo.csv

echo "=== git diff (cambio sin preparar) ==="
git diff

git add catalogo.csv
echo "=== git diff (ahora vacío: todo está preparado) ==="
git diff
echo "=== git diff --staged (lo que irá en el commit) ==="
git diff --staged

git commit -q -m "fix: actualizar precio del mouse"
echo "=== git diff HEAD~1 HEAD --stat (resumen del último commit) ==="
git diff HEAD~1 HEAD --stat
````

# Parte 6. Deshacer cosas

| Situación | Comando | ¿Es seguro? |
|---|---|---|
| Cambié un archivo y me arrepiento (aún no hice `add`) | `git restore archivo` | ⚠️ Descarta el cambio **para siempre** |

| Volver el historial atrás en **mi copia local** | `git reset --soft / --mixed / --hard` | ⚠️ `--hard` puede borrar trabajo |
| Perdí un commit tras un `reset` | `git reflog` | ✅ Es la red de seguridad |

## 6.4 `git reset` y `git reflog`: mover el historial (con cuidado)

`reset` **mueve la rama hacia atrás**, olvidando los commits posteriores. Tiene tres sabores, según qué hace con los cambios de esos commits:

| Modo | El commit | Los cambios quedan... |
|---|---|---|
| `--soft` | se deshace | **preparados** (en la caja) |
| `--mixed` (el modo por defecto) | se deshace | **en el directorio de trabajo**, sin preparar |
| `--hard` | se deshace | **se borran** ⚠️ |

La buena noticia es `git reflog`: Git anota **cada movimiento de `HEAD`** durante varias semanas (por defecto, entre 30 y 90 días) y eso permite recuperar un commit "perdido".

````python
%%bash
cd taller_git/s06

echo "Parlante portátil,Audio,95000" >> catalogo.csv
git commit -q -am "feat: agregar parlante (prueba)"

echo "=== 1) reset --soft HEAD~1: el commit se deshace, el cambio queda PREPARADO ==="
git reset -q --soft HEAD~1
git status --short catalogo.csv
git commit -q -m "feat: agregar parlante (prueba)"

echo "=== 2) reset --mixed HEAD~1: el cambio queda SIN preparar ==="
git reset -q --mixed HEAD~1
git status --short catalogo.csv
git commit -q -am "feat: agregar parlante (prueba)"

echo "=== 3) reset --hard HEAD~1: ¡el cambio se borra! ==="
git reset -q --hard HEAD~1
git status --short catalogo.csv
echo "Última línea del catálogo ahora:"
tail -n 1 catalogo.csv

echo "=== 4) git reflog: la red de seguridad ==="
git reflog -n 4
git reset -q --hard "HEAD@{1}"
echo "Última línea del catálogo tras recuperar:"
tail -n 1 catalogo.csv
````

| `venv/` | La carpeta `venv` completa |
| `__pycache__/` | Las carpetas `__pycache__` en cualquier nivel |
| `# texto` | Comentario (Git lo ignora) |

````python
%%bash
cat > taller_git/s07/.gitignore << 'EOF'
# Entornos virtuales
venv/
.venv/

# Python
__pycache__/
*.pyc

# Django
db.sqlite3
staticfiles/
media/

# Secretos
.env

# Editores
.vscode/
.idea/
EOF
echo "Archivo .gitignore creado:"
cat taller_git/s07/.gitignore
````

## 7.3 El error más común: el archivo ya estaba "seguido"

Un `.gitignore` solo afecta a archivos que Git **todavía no sigue**. Si un archivo ya fue guardado en un commit, Git seguirá vigilándolo aunque lo agregue al `.gitignore`. Para solucionarlo se le dice a Git que deje de seguirlo, **sin borrar el archivo del disco**: `git rm --cached`.

````python
%%bash
cd taller_git/s07

# Simulamos el error: db.sqlite3 se subió a la fuerza (-f) antes de que existiera la regla
git add -f db.sqlite3
git commit -q -m "feat: base de datos de prueba (error)"

echo "más datos" >> db.sqlite3
echo "--- Aunque está en .gitignore, Git sigue vigilándolo (M) ---"
git status --short

echo "--- Corrección: dejar de seguirlo, conservando el archivo en el disco ---"
git rm --cached -q db.sqlite3
git commit -q -m "chore: dejar de seguir db.sqlite3"
git status --short
echo "--- Archivos que Git sigue ahora ---"
git ls-files
echo "--- El archivo sigue en el disco ---"
ls db.sqlite3
````

> ⚠️ **Recuerde:** `git rm --cached` evita que el archivo se siga guardando, pero **las versiones anteriores siguen en el historial**. Si lo que se subió fue un secreto (una clave o una contraseña), considere que ya quedó expuesto y **cámbielo**.

# Parte 8. Ramas

## 8.1 ¿Para qué sirven?

Imagine que TiendaTech Cali ya tiene su catálogo funcionando y usted quiere probar un sistema de descuentos. Si lo programa directamente sobre el código que funciona, un error puede dañarlo todo. Con Git usted abre una **rama**: una línea de trabajo paralela donde puede experimentar. Si sale bien, la **fusiona** (*merge*) con la rama principal; si sale mal, la descarta y la rama principal queda intacta.

Recuerde la Parte 3: una rama es solo un archivo con la huella de un commit, por eso crearlas es instantáneo.

| Comando | Para qué |
|---|---|
| `git branch` | Listar ramas (la activa lleva `*`) |
| `git switch -c nombre` | Crear una rama y cambiarse a ella |
| `git switch nombre` | Cambiarse a una rama existente |
| `git merge nombre` | Fusionar la rama `nombre` en la rama activa |
| `git branch -d nombre` | Borrar una rama ya fusionada |

En versiones antiguas de Git verá `git checkout -b nombre` y `git checkout nombre`; hacen lo mismo que `switch`.

## 8.2 Crear una rama, trabajar en ella y fusionarla

Observe cómo, al volver a `main`, el archivo de la rama **desaparece** de la carpeta (¡Git cambia los archivos según la rama!) y **reaparece** al fusionar.

````python
%%bash
rm -rf taller_git/s08; mkdir -p taller_git/s08; cd taller_git/s08
git init -q -b main
printf "producto,categoria,precio\nLaptop Gamer,Computadores,4500000\n" > catalogo.csv
git add catalogo.csv
git commit -q -m "feat: agregar catálogo inicial"
echo "--- Ramas existentes ---"
git branch

echo "--- Crear la rama feature/descuentos y trabajar en ella ---"
git switch -c feature/descuentos
printf "def descuento(precio, porcentaje):\n    return precio - precio * porcentaje / 100\n" > descuentos.py
git add descuentos.py
git commit -q -m "feat: agregar función de descuento"
git log --oneline --graph --all

echo "--- Volver a main: descuentos.py desaparece ---"
git switch main
ls

echo "--- Fusionar la rama en main ---"
git merge feature/descuentos
ls
git log --oneline --graph --all

echo "--- La rama ya cumplió su función: se borra ---"
git branch -d feature/descuentos
git branch
````

Esa fusión fue de tipo **fast-forward** ("avance rápido"): como `main` no había cambiado mientras usted trabajaba en la rama, Git solo tuvo que mover el puntero de `main` hacia adelante. No hizo falta ningún commit extra.

## 8.3 Cuando las dos ramas avanzaron: commit de fusión

Si `main` también recibió commits mientras la rama avanzaba, Git no puede simplemente mover el puntero: crea un **commit de fusión** (*merge commit*) con **dos padres**. En el dibujo del `log --graph` verá las líneas que se separan y se vuelven a unir.

````python
%%bash
cd taller_git/s08

git switch -q -c feature/ofertas
echo "# Ofertas de temporada" > ofertas.md
git add ofertas.md
git commit -q -m "docs: agregar ofertas"

git switch -q main
echo "Mouse inalámbrico,Perifericos,55000" >> catalogo.csv
git commit -q -am "feat: agregar mouse inalámbrico"

echo "--- Las dos ramas avanzaron por separado ---"
git log --oneline --graph --all

echo "--- Fusión con commit de fusión ---"
git merge --no-edit feature/ofertas
git log --oneline --graph --all
git branch -d feature/ofertas
````

> **Dato útil:** la opción `--no-ff` obliga a Git a crear el commit de fusión **aunque pudiera avanzar rápido**. Muchos equipos la usan para que en el historial quede siempre registro de que una funcionalidad existió como rama.

# Parte 9. Conflictos de fusión

## 9.1 ¿Qué es un conflicto y por qué no hay que temerle?

Un **conflicto** ocurre cuando dos ramas cambiaron **la misma línea del mismo archivo** de maneras distintas. Git no puede adivinar cuál versión es la correcta, así que se detiene y le pide a usted que decida. **No es un error: es Git pidiendo una decisión humana.**

Un ejemplo de la vida real: una persona del equipo sube el precio del mouse a 59000 y otra, en una rama de ofertas, lo baja a 49000. ¿Cuál queda?

## 9.2 Provocar un conflicto y leerlo

````python
%%bash
rm -rf taller_git/s09; mkdir -p taller_git/s09; cd taller_git/s09
git init -q -b main
printf "producto,categoria,precio\nLaptop Gamer,Computadores,4500000\nMouse inalámbrico,Perifericos,55000\n" > catalogo.csv
git add catalogo.csv
git commit -q -m "feat: agregar catálogo inicial"

git switch -q -c oferta-mouse
reemplazar 55000 49000 catalogo.csv
git commit -q -am "feat: oferta del mouse"

git switch -q main
reemplazar 55000 59000 catalogo.csv
git commit -q -am "fix: subir precio del mouse"

echo "--- Intentamos fusionar la rama de oferta en main ---"
git merge oferta-mouse || echo ">>> Conflicto detectado (es lo esperado en este ejemplo)"
echo "--- git status: el archivo aparece como UU (modificado por ambos) ---"
git status --short
echo "--- Así quedó el archivo: Git dejó marcas para que usted decida ---"
cat catalogo.csv
````

### Cómo leer las marcas

```
<<<<<<< HEAD                          ← comienza TU versión (la rama en la que estás: main)
Mouse inalámbrico,Perifericos,59000
=======                               ← separador
Mouse inalámbrico,Perifericos,49000
>>>>>>> oferta-mouse                  ← termina la versión de la OTRA rama
```

## 9.3 Resolver el conflicto: 4 pasos

1. **Abra el archivo** y decida qué versión queda (la suya, la otra, o una combinación).
2. **Borre las marcas** `<<<<<<<`, `=======` y `>>>>>>>` junto con lo que sobre. El archivo debe quedar como usted quiere que sea.
3. **Prepare** el archivo con `git add` (esto le dice a Git "ya lo resolví").
4. **Termine la fusión** con `git commit`.

> 💡 **En VS Code** el trabajo del paso 1 y 2 es más cómodo: al abrir un archivo en conflicto, el editor resalta las dos versiones y muestra, sobre cada conflicto, los botones **Accept Current Change** (aceptar mi versión), **Accept Incoming Change** (aceptar la de la otra rama), **Accept Both Changes** (aceptar ambas) y **Compare Changes** (compararlas). ## 9.4 Atajos y salidas de emergencia

| Situación | Comando |
|---|---|
| Quiero quedarme con **mi** versión del archivo completo | `git checkout --ours archivo` |
| Quiero quedarme con la versión de **la otra rama** del archivo completo | `git checkout --theirs archivo` |
| Me asusté y quiero **cancelar** la fusión como si nada | `git merge --abort` |

> ⚠️ `--ours` y `--theirs` reemplazan **todo el archivo** por una de las dos versiones, no solo las líneas en conflicto. Si el archivo tenía otros cambios no conflictivos del otro lado, se perderán. Úselos cuando tenga claro que quiere una versión completa.
>
> Después de `--ours` o `--theirs`, igualmente hay que hacer `git add archivo` y `git commit`.

Veamos la salida de emergencia: otra fusión conflictiva, pero esta vez la **cancelamos**.

# Parte 10. Repositorios remotos: `push`, `pull`, `fetch` y `clone`

## 10.1 ¿Qué es un remoto?

Hasta ahora todo el trabajo ocurrió en un solo repositorio. Para colaborar necesitamos un lugar compartido al que cada persona **suba** y del que **descargue**. Ese lugar es un **repositorio remoto**. GitHub, GitLab y Bitbucket son servicios que alojan repositorios remotos; técnicamente, un remoto es solo otro repositorio, normalmente "sin carpeta de trabajo" (un repositorio *bare*), donde todos hacen `push` y `pull`.

Para practicar **sin crear cuentas**, simularemos el servidor con una carpeta de su propio computador: es exactamente el mismo mecanismo, solo que la dirección es una ruta de carpeta en vez de una dirección de internet.

```
   Su repositorio                      Remoto ("GitHub")                  Repositorio de su compañera
   (carpeta yo/)        ───── push ────▶ (carpeta remoto/)  ◀──── push ─────  (carpeta companero/)
                        ◀──── pull ─────                     ───── pull ────▶
```

| Comando | Qué hace |
|---|---|
| `git remote add origin <dirección>` | Registra el remoto con el nombre `origin` (es solo un apodo) |
| `git push -u origin main` | Sube la rama `main` y la deja "enlazada" con la del remoto |
| `git clone <dirección>` | Descarga una copia completa (con historial) y ya queda conectada al remoto |
| `git fetch` | Mira qué hay de nuevo en el remoto, **sin tocar** sus archivos |
| `git pull` | `fetch` + `merge`: trae los cambios y los fusiona en su rama |

## 10.2 Crear el remoto y subir el primer commit

````python
%%bash
rm -rf taller_git/s10; mkdir -p taller_git/s10/remoto taller_git/s10/yo
RAIZ="$(pwd)/taller_git/s10"

echo "--- 1) El 'servidor' (en GitHub esto lo hace el botón New repository) ---"
git init -q --bare -b main "$RAIZ/remoto/tienda.git"

echo "--- 2) Su repositorio local ---"
cd "$RAIZ/yo"
mkdir tienda
cd tienda
git init -q -b main
printf "producto,categoria,precio\nLaptop Gamer,Computadores,4500000\n" > catalogo.csv
git add catalogo.csv
git commit -q -m "feat: agregar catálogo inicial"

echo "--- 3) Conectar ambos: el remoto se llama 'origin' ---"
git remote add origin "$RAIZ/remoto/tienda.git"
git remote -v
git push -u origin main
echo "--- La rama main ahora sigue a origin/main ---"
git branch -vv
````

## 10.3 Una compañera clona el repositorio y sube un cambio

`git clone` descarga todo. Para diferenciar a las personas en el historial, la compañera se identifica con otro nombre **solo en su copia** (`git config` sin `--global`).

````python
%%bash
RAIZ="$(pwd)/taller_git/s10"
mkdir -p "$RAIZ/companero"
git clone -q "$RAIZ/remoto/tienda.git" "$RAIZ/companero/tienda"
cd "$RAIZ/companero/tienda"
git config user.name "Compañera de equipo"
git config user.email "companera@ejemplo.com"
echo "# TiendaTech Cali" > README.md
git add README.md
git commit -q -m "docs: agregar README"
git push -q origin main
git log --format="%h  %an: %s"
````

## 10.4 `fetch` y `pull`: ponerse al día

Su copia todavía no sabe que su compañera subió algo. `git fetch` pregunta al remoto qué hay de nuevo **sin modificar sus archivos**; así usted puede mirar antes de decidir. `git pull` hace ese paso y además fusiona.

En `git status -sb`, la palabra `behind` ("atrás") le avisa de cuántos commits le faltan.

````python
%%bash
RAIZ="$(pwd)/taller_git/s10"
cd "$RAIZ/yo/tienda"
echo "--- Su historial, antes de enterarse de lo de su compañera ---"
git log --oneline
git fetch -q
echo "--- Después de fetch: usted ve que va 1 commit atrás ---"
git status -sb | head -n 1
git log --oneline origin/main
echo "--- git pull: trae y fusiona ---"
git pull --no-edit
git log --format="%h  %an: %s"
````

## 10.5 Cuando el `push` es rechazado

Si su compañera sube un cambio **antes que usted**, el remoto tendrá commits que usted no tiene, y Git rechazará su `push` para no pisar su trabajo. La solución siempre es la misma: **primero traiga los cambios (`pull`), luego suba.**

````python
%%bash
RAIZ="$(pwd)/taller_git/s10"

# Usted hace un cambio local (sin subirlo todavía)
cd "$RAIZ/yo/tienda"
echo "Teclado mecánico,Perifericos,180000" >> catalogo.csv
git commit -q -am "feat: agregar teclado mecánico"

# Su compañera, mientras tanto, sube otro cambio
cd "$RAIZ/companero/tienda"
git pull -q --no-edit
echo "Uso educativo" > LICENSE.md
git add LICENSE.md
git commit -q -m "docs: agregar licencia"
git push -q origin main

# Ahora usted intenta subir
cd "$RAIZ/yo/tienda"
echo "--- Usted intenta subir: el remoto lo rechaza ---"
git push origin main || echo ">>> Push rechazado (es lo esperado en este ejemplo)"
echo "--- Solución: primero traer los cambios (pull) y luego subir ---"
git pull --no-edit
git push -q origin main
git log --oneline --graph
````

## 11.2 `stash`: guardar trabajo a medias

Está trabajando en algo sin terminar y le piden arreglar algo urgente. No quiere hacer un commit de trabajo incompleto ni perder lo escrito. `git stash` guarda sus cambios en un "cajón" y deja la carpeta limpia; luego `git stash pop` los saca de nuevo.

| Comando | Qué hace |
|---|---|
| `git stash push -m "nota"` | Guarda los cambios pendientes y limpia la carpeta |
| `git stash list` | Muestra lo guardado en el cajón |
| `git stash pop` | Vuelve a aplicar lo último guardado y lo saca del cajón |

````python
%%bash
cd taller_git/s11
echo "Webcam HD,Perifericos,130000" >> catalogo.csv
echo "--- Trabajo a medias en la carpeta ---"
git status --short
git stash push -m "webcam a medias"
echo "--- Carpeta limpia (ya puede atender lo urgente) ---"
git status --short
git stash list
git stash pop -q
echo "--- Trabajo recuperado en la carpeta ---"
git status --short
echo "--- Contenido del cajón tras el pop (no debe mostrar nada) ---"
git stash list
git restore catalogo.csv
````

## 12.2 Flujo de trabajo en equipo: *GitHub Flow*

La rama `main` siempre debe estar estable. Cada cambio nuevo se hace en una rama propia y entra a `main` mediante una **Pull Request**.

```
main ──●────────────────────────●──────────●──   (siempre estable)
        \                      /
         ●───●───●────────────●                  feature/descuentos (su rama de trabajo)
                              │
                     Pull Request (revisión del equipo)
```

1. Parta de `main` actualizada: `git switch main` y `git pull`.
2. Cree una rama con un nombre claro: `git switch -c feature/descuentos` (o `fix/precio-mouse`).
3. Haga **commits pequeños** con buenos mensajes.
4. Suba la rama: `git push -u origin feature/descuentos`.
5. Abra una **Pull Request** en GitHub: es una solicitud para fusionar su rama en `main`, donde sus compañeros revisan el código, comentan y aprueban.
6. Fusione la Pull Request desde GitHub y borre la rama.
7. En su computador: `git switch main`, `git pull` y `git branch -d feature/descuentos`.

**Reglas de buena convivencia:** no se hacen commits directos a `main`; cada Pull Request resuelve **una** cosa; los mensajes de commit siguen la convención de la Parte 4.

## 12.3 ¿Qué se sube y qué no, en un proyecto Django?

| Se sube ✅ | No se sube ❌ |
|---|---|
| Código (`.py`, plantillas `.html`, archivos estáticos propios) | `venv/` o `.venv/` (el entorno virtual) |
| `requirements.txt` (lista de librerías) | `db.sqlite3` (base de datos local) |
| Migraciones (`migrations/*.py`) | `.env` (secretos y `SECRET_KEY`) |
| `README.md` y `.gitignore` | `__pycache__/` y archivos `.pyc` |
| Un `.env.example` **sin** valores reales | `staticfiles/` y `media/` generados |
| | Archivos de registro `.log` y carpetas del editor (`.idea/`) |

La regla de oro: **si un archivo se puede regenerar o contiene un secreto, no se sube.** Quien clone el proyecto crea su propio entorno virtual con `pip install -r requirements.txt` y su propio `.env`.

## 13.2 Errores frecuentes de Git y qué hacer

| Mensaje de Git | Qué significa | Qué hacer |
|---|---|---|
| `fatal: not a git repository` | Está en una carpeta que no es un repositorio | Entre a la carpeta correcta (`cd`) o ejecute `git init` |
| `Author identity unknown` / `Please tell me who you are` | Git no sabe su nombre y correo | Ejecute la celda 0.3 o `git config --global user.name "..."` y `user.email "..."` |
| `nothing to commit, working tree clean` | No hay cambios por guardar | No es un error: ya está todo guardado |
| `nothing added to commit but untracked files present` | Hay archivos nuevos sin preparar | Use `git add archivo` antes de `commit` |
| `error: Your local changes would be overwritten by...` | Tiene cambios sin guardar que chocan con lo que va a traer | Haga commit de sus cambios, o use `git stash` |
| `! [rejected] ... (non-fast-forward)` | El remoto tiene commits que usted no tiene | `git pull` y luego `git push` |
| `CONFLICT (content): Merge conflict in ...` | Dos ramas cambiaron la misma línea | Resuelva (Parte 9) o cancele con `git merge --abort` |
| `You are in 'detached HEAD' state` | Está mirando un commit antiguo, no una rama | Vuelva con `git switch main`; si quiere conservar lo hecho, `git switch -c nueva-rama` |
| `fatal: refusing to merge unrelated histories` | Dos repositorios sin pasado en común | Verifique que clonó el correcto; si es intencional, `git pull --allow-unrelated-histories` |
| `warning: LF will be replaced by CRLF` (Windows) | Diferencia de saltos de línea entre sistemas | Es solo un aviso; para silenciarlo: `git config --global core.autocrlf true` |
| Subí un `.env` o una clave por error | Un secreto quedó en el historial | **Cambie la clave de inmediato**, luego `git rm --cached` y agréguelo al `.gitignore` |

````python
T = "taller_git/taller"
LOCAL = T + "/catalogo"
REMOTO = T + "/remoto/catalogo.git"

def archivos_en_head(carpeta):
    return git_texto(carpeta, "ls-tree", "-r", "--name-only", "HEAD").splitlines()

def asuntos(carpeta, *extra):
    return git_texto(carpeta, "log", *extra, "--format=%s").splitlines()

def leer(ruta):
    try:
        with open(ruta, encoding="utf-8") as archivo:
            return archivo.read()
    except OSError:
        return ""

def v1():
    print("Misión 1 · Preparar el repositorio")
    r = [comprobar("La rama activa se llama main", git_texto(LOCAL, "branch", "--show-current") == "main"),
         comprobar("Existe el commit 'docs: crear README del catálogo'", "docs: crear README del catálogo" in asuntos(LOCAL)),
         comprobar("README.md está en el historial", "README.md" in archivos_en_head(LOCAL)),
         comprobar("Nombre y correo configurados solo en este repositorio",
                   git_texto(LOCAL, "config", "--local", "user.name") != ""
                   and git_texto(LOCAL, "config", "--local", "user.email") != "")]
    return all(r)

def v2():
    print("Misión 2 · Proteger lo sensible con .gitignore")
    r = [comprobar("Existe el commit 'feat: agregar catálogo inicial'", "feat: agregar catálogo inicial" in asuntos(LOCAL)),
         comprobar("Existe el commit 'chore: agregar .gitignore'", "chore: agregar .gitignore" in asuntos(LOCAL)),
         comprobar("catalogo.csv y .gitignore están en el historial",
                   "catalogo.csv" in archivos_en_head(LOCAL) and ".gitignore" in archivos_en_head(LOCAL)),
         comprobar(".env NO está en el historial y es ignorado",
                   ".env" not in archivos_en_head(LOCAL) and git_texto(LOCAL, "check-ignore", ".env") == ".env"),
         comprobar("servidor.log NO está en el historial y es ignorado",
                   "servidor.log" not in archivos_en_head(LOCAL) and git_texto(LOCAL, "check-ignore", "servidor.log") == "servidor.log")]
    return all(r)

def v3():
    print("Misión 3 · Rama de descuentos y fusión")
    r = [comprobar("Existe el commit 'feat: agregar función de descuento'", "feat: agregar función de descuento" in asuntos(LOCAL)),
         comprobar("descuentos.py está en main", "descuentos.py" in archivos_en_head(LOCAL)),
         comprobar("Hay un commit de fusión 'merge: incorporar descuentos'",
                   "merge: incorporar descuentos" in asuntos(LOCAL, "--merges")),
         comprobar("La rama feature/descuentos fue borrada", git_texto(LOCAL, "branch", "--list", "feature/descuentos") == "")]
    return all(r)

def v4():
    print("Misión 4 · Resolver un conflicto")
    texto = leer(LOCAL + "/catalogo.csv")
    r = [comprobar("El archivo no tiene marcas de conflicto", texto != "" and "<<<<<<<" not in texto and ">>>>>>>" not in texto),
         comprobar("La laptop gamer quedó en 4300000 (precio de la rama)",
                   "Laptop Gamer,Computadores,4300000" in texto.splitlines()),
         comprobar("Hay un commit de fusión de feature/precio-gamer",
                   any("precio-gamer" in s for s in asuntos(LOCAL, "--merges"))),
         comprobar("No queda nada pendiente", git_texto(LOCAL, "status", "--short") == "")]
    return all(r)

def v5():
    print("Misión 5 · Colaborar a través de un remoto")
    r = [comprobar("El remoto se llama origin", git_texto(LOCAL, "remote") == "origin"),
         comprobar("El remoto recibió la rama main", git_texto(REMOTO, "rev-parse", "main") != ""),
         comprobar("El commit de su compañera está en el remoto",
                   "feat: agregar webcam HD" in git_texto(REMOTO, "log", "main", "--format=%s").splitlines()),
         comprobar("La webcam ya está en su catálogo local",
                   "Webcam HD,Perifericos,130000" in leer(LOCAL + "/catalogo.csv").splitlines())]
    return all(r)

def v6():
    print("Misión 6 · Deshacer un error ya compartido")
    r = [comprobar("credenciales.txt ya NO está en main", "credenciales.txt" not in archivos_en_head(LOCAL)),
         comprobar("La historia conserva un commit Revert", any(s.startswith("Revert") for s in asuntos(LOCAL))),
         comprobar("El remoto tiene la corrección (HEAD local == main del remoto)",
                   git_texto(LOCAL, "rev-parse", "HEAD") != ""
                   and git_texto(LOCAL, "rev-parse", "HEAD") == git_texto(REMOTO, "rev-parse", "main"))]
    return all(r)

def v7():
    print("Misión 7 · Etiquetar la versión 1.0.0")
    r = [comprobar("v1.0.0 existe y es una etiqueta anotada", git_texto(LOCAL, "cat-file", "-t", "v1.0.0") == "tag"),
         comprobar("Tiene el mensaje 'Primera versión del catálogo'",
                   "Primera versión del catálogo" in git_texto(LOCAL, "tag", "-n1", "v1.0.0")),
         comprobar("La etiqueta fue subida al remoto", "v1.0.0" in git_texto(REMOTO, "tag").splitlines())]
    return all(r)

print("Verificaciones del taller listas.")
````
